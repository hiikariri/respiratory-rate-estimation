## Respiration Signal

```python
import numpy as np
from scipy.signal import find_peaks

# Extract respiration signal
resp_signal = data['RESP'].values
time_signal = data['Time [s]'].values

# Load breath annotations from CSV file
breath_data = pd.read_csv(f'data/bidmc-ppg-and-respiration-dataset/bidmc_csv/{data_path}Breaths.csv')

# Extract breath sample numbers from both annotators
breath_samples_ann1 = breath_data.iloc[:, 0].values  # First annotator
breath_samples_ann2 = breath_data.iloc[:, 1].values  # Second annotator

# Use average of both annotators for ground truth
breath_samples = ((breath_samples_ann1 + breath_samples_ann2) / 2).astype(int)

print(f"Loaded {len(breath_samples)} annotated breaths")
print(f"First 10 breath samples: {breath_samples[:10]}")

# Segment data into 60-second windows
segment_duration = 60  # seconds
total_duration = time_signal[-1] - time_signal[0]
num_segments = int(np.ceil(total_duration / segment_duration))

print(f"\nTotal duration: {total_duration:.2f} seconds")
print(f"Number of 60-second segments: {num_segments}\n")

# Calculate sampling frequency
sampling_freq = 1 / (time_signal[1] - time_signal[0])

# Store results for each segment
segment_results = []

for i in range(num_segments):
    # Define segment time range
    start_time = i * segment_duration
    end_time = min((i + 1) * segment_duration, time_signal[-1])
    
    # Convert time to sample indices
    start_sample = int(start_time * sampling_freq)
    end_sample = int(end_time * sampling_freq)
    
    # Count annotated breaths within this segment
    num_breaths = np.sum((breath_samples >= start_sample) & (breath_samples <= end_sample))
    
    # Calculate respiratory rate for this segment
    segment_dur = end_time - start_time
    respiratory_rate = (num_breaths / segment_dur) * 60 if segment_dur > 0 else 0
    
    segment_results.append({
        'segment': i + 1,
        'start_time': start_time,
        'end_time': end_time,
        'num_breaths': num_breaths,
        'respiratory_rate': respiratory_rate
    })
    
    print(f"Segment {i+1}: {start_time:.1f}s - {end_time:.1f}s")
    print(f"  Annotated breaths: {num_breaths}")
    print(f"  Respiratory Rate: {respiratory_rate:.2f} BPM\n")

# Create a DataFrame for easy viewing
segments_df = pd.DataFrame(segment_results)
print("\nSummary Statistics:")
print(f"Mean Respiratory Rate: {segments_df['respiratory_rate'].mean():.2f} BPM")
print(f"Std Respiratory Rate: {segments_df['respiratory_rate'].std():.2f} BPM")
print(f"Min Respiratory Rate: {segments_df['respiratory_rate'].min():.2f} BPM")
print(f"Max Respiratory Rate: {segments_df['respiratory_rate'].max():.2f} BPM")

# Visualize respiration signal with annotated breaths
fig = go.Figure()

# Add respiration signal
fig.add_trace(go.Scatter(
    x=time_signal, 
    y=resp_signal,
    mode='lines',
    name='RESP Signal',
    line=dict(color='blue', width=1)
))

# Add annotated breaths as markers
breath_times = breath_samples / sampling_freq
# Filter to valid range
valid_breath_mask = (breath_times >= time_signal[0]) & (breath_times <= time_signal[-1])
valid_breath_times = breath_times[valid_breath_mask]
valid_breath_samples = breath_samples[valid_breath_mask]

# Get corresponding signal values
breath_signal_values = resp_signal[valid_breath_samples]

fig.add_trace(go.Scatter(
    x=valid_breath_times,
    y=breath_signal_values,
    mode='markers',
    name='Annotated Breaths',
    marker=dict(color='red', size=6, symbol='x')
))

# Add vertical lines to show segment boundaries
for i in range(1, num_segments):
    fig.add_vline(x=i * segment_duration, line_dash="dash", line_color="gray", opacity=0.5)

fig.update_layout(
    title=f'Respiration Signal with Annotated Breaths (60s Segments)',
    xaxis_title='Time [s]',
    yaxis_title='RESP',
    height=600,
    showlegend=True
)

fig.show()

# Display the segments DataFrame
print("\n" + "="*50)
print("Segment Details:")
print(segments_df.to_string(index=False))
```

## ECG Peak Detection

```python
from scipy.signal import butter, filtfilt
import pywt

# Extract ECG signal (Lead II)
ecg_signal = data['II'].values
time_signal = data['Time [s]'].values

# Calculate sampling frequency
sampling_freq = 1 / (time_signal[1] - time_signal[0])
print(f"Sampling frequency: {sampling_freq:.2f} Hz\n")

# Design and apply 4th order Butterworth bandpass filter (5-40 Hz)
def bandpass_filter(signal, lowcut, highcut, fs, order=4):
    nyquist = 0.5 * fs
    low = lowcut / nyquist
    high = highcut / nyquist
    b, a = butter(order, [low, high], btype='band')
    filtered_signal = filtfilt(b, a, signal)
    return filtered_signal

# Apply bandpass filter
ecg_filtered = bandpass_filter(ecg_signal, lowcut=5, highcut=40, fs=sampling_freq, order=4)
print("ECG signal filtered with 4th order BPF (5-40 Hz)\n")

# Segment data into 60-second windows
segment_duration = 60  # seconds
total_duration = time_signal[-1] - time_signal[0]
num_segments = int(np.ceil(total_duration / segment_duration))

print(f"Total duration: {total_duration:.2f} seconds")
print(f"Number of 60-second segments: {num_segments}\n")

# Function to detect R peaks using Pan-Tompkins Algorithm
def detect_r_peaks_pan_tompkins(ecg_segment, fs):
    """
    Detect R peaks using Pan-Tompkins Algorithm
    """
    # Step 1: Bandpass filter (5-15 Hz) - already done in preprocessing, but apply again for robustness
    filtered = bandpass_filter(ecg_segment, lowcut=5, highcut=15, fs=fs, order=4)
    
    # Step 2: Derivative
    derivative = np.zeros_like(filtered)
    derivative[1:] = filtered[1:] - filtered[:-1]
    
    # Step 3: Square
    squared = derivative ** 2
    
    # Step 4: Moving window integration
    window_size = int(0.150 * fs)  # 150ms window
    mwa = np.zeros_like(squared)
    half_window = window_size // 2
    for i in range(len(squared)):
        start = max(0, i - half_window)
        end = min(len(squared), i + half_window + 1)
        mwa[i] = np.mean(squared[start:end])
    
    # Step 5: Thresholding and peak detection
    threshold = np.mean(mwa)
    # threshold = 0.35 * np.max(mwa)
    min_distance = int(0.2 * fs)  # Minimum 200ms between peaks
    approx_peaks = []
    last_peak = -min_distance
    
    for i in range(1, len(mwa) - 1):
        if mwa[i] > threshold and mwa[i] > mwa[i-1] and mwa[i] > mwa[i+1]:
            if i - last_peak >= min_distance:
                approx_peaks.append(i)
                last_peak = i
    
    # Step 6: Find actual R peaks in original ECG
    refined_peaks = []
    search_width = int(0.1 * fs)  # 100ms search window
    for peak in approx_peaks:
        start = max(0, peak - search_width)
        end = min(len(ecg_segment), peak + search_width)
        local_max_idx = start + np.argmax(ecg_segment[start:end])
        refined_peaks.append(local_max_idx)
    
    return np.array(refined_peaks)

# Store results for each segment
ecg_segment_results = []
ecg_segment_peaks = []  # Store R peaks for visualization

for i in range(num_segments):
    # Define segment time range
    start_time = i * segment_duration
    end_time = min((i + 1) * segment_duration, time_signal[-1])
    
    # Find indices for this segment
    segment_mask = (time_signal >= start_time) & (time_signal < end_time)
    segment_time = time_signal[segment_mask]
    segment_ecg = ecg_filtered[segment_mask]
    
    if len(segment_ecg) > 0:
        # Detect R peaks using pan tompkins
        r_peaks = detect_r_peaks_pan_tompkins(segment_ecg, sampling_freq)
        
        # Calculate heart rate for this segment
        segment_dur = segment_time[-1] - segment_time[0]
        num_beats = len(r_peaks)
        heart_rate = (num_beats / segment_dur) * 60 if segment_dur > 0 else 0
        
        # Store results
        ecg_segment_results.append({
            'segment': i + 1,
            'start_time': start_time,
            'end_time': end_time,
            'num_beats': num_beats,
            'heart_rate': heart_rate
        })
        
        # Store peaks with global indices for visualization
        if len(r_peaks) > 0:
            global_peak_indices = np.where(segment_mask)[0][r_peaks]
            ecg_segment_peaks.append((i, global_peak_indices))
        
        print(f"Segment {i+1}: {start_time:.1f}s - {end_time:.1f}s")
        print(f"  R peaks detected: {num_beats}")
        print(f"  Heart Rate: {heart_rate:.2f} BPM\n")

# Create a DataFrame for easy viewing
ecg_segments_df = pd.DataFrame(ecg_segment_results)
print("\nECG Summary Statistics:")
print(f"Mean Heart Rate: {ecg_segments_df['heart_rate'].mean():.2f} BPM")
print(f"Std Heart Rate: {ecg_segments_df['heart_rate'].std():.2f} BPM")
print(f"Min Heart Rate: {ecg_segments_df['heart_rate'].min():.2f} BPM")
print(f"Max Heart Rate: {ecg_segments_df['heart_rate'].max():.2f} BPM")

# Visualize filtered ECG with detected R peaks
fig = make_subplots(
    rows=2, cols=1,
    subplot_titles=('Original ECG Signal', 'Filtered ECG with R Peaks Detected'),
    vertical_spacing=0.1
)

# Add original ECG signal
fig.add_trace(go.Scatter(
    x=time_signal, 
    y=ecg_signal,
    mode='lines',
    name='Original ECG',
    line=dict(color='gray', width=0.5)
), row=1, col=1)

# Add filtered ECG signal
fig.add_trace(go.Scatter(
    x=time_signal, 
    y=ecg_filtered,
    mode='lines',
    name='Filtered ECG',
    line=dict(color='blue', width=1)
), row=2, col=1)

# Add R peaks for each segment with different colors
colors = ['red', 'green', 'orange', 'purple', 'brown', 'pink', 'gray', 'olive', 'cyan']

for segment_idx, peak_indices in ecg_segment_peaks:
    fig.add_trace(go.Scatter(
        x=time_signal[peak_indices],
        y=ecg_filtered[peak_indices],
        mode='markers',
        name=f'Segment {segment_idx + 1} R Peaks',
        marker=dict(color=colors[segment_idx % len(colors)], size=6, symbol='x'),
        showlegend=(segment_idx < 3)  # Only show first 3 in legend to avoid clutter
    ), row=2, col=1)

# Add vertical lines to show segment boundaries
for i in range(1, num_segments):
    fig.add_vline(x=i * segment_duration, line_dash="dash", line_color="gray", opacity=0.3, row=1, col=1)
    fig.add_vline(x=i * segment_duration, line_dash="dash", line_color="gray", opacity=0.3, row=2, col=1)

fig.update_xaxes(title_text='Time [s]', row=2, col=1)
fig.update_yaxes(title_text='ECG', row=1, col=1)
fig.update_yaxes(title_text='Filtered ECG', row=2, col=1)

fig.update_layout(
    height=800,
    title_text='ECG Signal Processing and R Peak Detection (DWT)'
)

fig.show()

# Display the segments DataFrame
print("\n" + "="*50)
print("ECG Segment Details:")
print(ecg_segments_df.to_string(index=False))
```

```python
from scipy.interpolate import CubicSpline

# Extract R peaks location (R_loc) and amplitude (R_amp) from all segments
R_loc = []  # R peak time locations
R_amp = []  # R peak amplitudes

for segment_idx, peak_indices in ecg_segment_peaks:
    # Get time locations
    peak_times = time_signal[peak_indices]
    
    # Get amplitudes from filtered ECG signal
    peak_amplitudes = ecg_filtered[peak_indices]
    
    # Append to arrays
    R_loc.extend(peak_times)
    R_amp.extend(peak_amplitudes)

# Convert to numpy arrays
R_loc = np.array(R_loc)
R_amp = np.array(R_amp)

print("="*70)
print("R PEAK EXTRACTION FOR EDR")
print("="*70)
print(f"Total R peaks detected: {len(R_loc)}")
print(f"Time range: {R_loc.min():.2f}s - {R_loc.max():.2f}s")
print(f"\nR_amp Statistics:")
print(f"  Mean: {R_amp.mean():.4f}")
print(f"  Std: {R_amp.std():.4f}")
print(f"  Min: {R_amp.min():.4f}")
print(f"  Max: {R_amp.max():.4f}")

# Create cubic spline interpolation for smooth EDR signal
cs = CubicSpline(R_loc, R_amp)

# Generate continuous time points for EDR signal
# Use the same time resolution as original signal
edr_time = np.arange(R_loc[0], R_loc[-1], time_signal[1] - time_signal[0])
edr_signal = cs(edr_time)

print(f"\nEDR Signal Generated:")
print(f"  Length: {len(edr_signal)} samples")
print(f"  Time range: {edr_time[0]:.2f}s - {edr_time[-1]:.2f}s")
print(f"  Duration: {edr_time[-1] - edr_time[0]:.2f}s")

# Normalize EDR signal and Resp Signal using z-score normalization
edr_signal = (edr_signal - np.mean(edr_signal)) / np.std(edr_signal)
resp_signal = (resp_signal - np.mean(resp_signal)) / np.std(resp_signal)

# MAV the EDR signal
# edr_signal = np.convolve(edr_signal, np.ones(5)/5, mode='same')

# Visualize R peaks and EDR signal
fig = make_subplots(
    rows=3, cols=1,
    subplot_titles=(
        'ECG Signal with R Peaks',
        'R Peak Amplitudes (R_amp)',
        'EDR Signal (Cubic Spline Interpolation)'
    ),
    vertical_spacing=0.1
)

# Plot 1: ECG with R peaks
fig.add_trace(go.Scatter(
    x=time_signal,
    y=ecg_filtered,
    mode='lines',
    name='Filtered ECG',
    line=dict(color='lightblue', width=0.8)
), row=1, col=1)

fig.add_trace(go.Scatter(
    x=R_loc,
    y=R_amp,
    mode='markers',
    name='R Peaks',
    marker=dict(color='red', size=5, symbol='x')
), row=1, col=1)

# Plot 2: Discrete R peak amplitudes
fig.add_trace(go.Scatter(
    x=R_loc,
    y=R_amp,
    mode='markers+lines',
    name='R_amp (discrete)',
    marker=dict(color='red', size=6),
    line=dict(color='red', width=1, dash='dot')
), row=2, col=1)

# Plot 3: Continuous EDR signal
fig.add_trace(go.Scatter(
    x=edr_time,
    y=edr_signal,
    mode='lines',
    name='EDR Signal',
    line=dict(color='green', width=2)
), row=3, col=1)

# Add original respiration signal for comparison
fig.add_trace(go.Scatter(
    x=time_signal,
    y=resp_signal,
    mode='lines',
    name='Original RESP',
    line=dict(color='blue', width=1, dash='dash'),
    opacity=0.5
), row=3, col=1)

# Update axes
fig.update_xaxes(title_text='Time [s]', row=3, col=1)
fig.update_yaxes(title_text='ECG Amplitude', row=1, col=1)
fig.update_yaxes(title_text='R_amp', row=2, col=1)
fig.update_yaxes(title_text='Amplitude', row=3, col=1)

fig.update_layout(
    height=1000,
    title_text='ECG-Derived Respiration (EDR) Signal Extraction',
    showlegend=True
)

fig.show()

# Display first 20 R peak values
print("\n" + "="*70)
print("First 20 R Peaks:")
print("="*70)
print(f"{'Index':<8} {'R_loc (s)':<15} {'R_amp':<15}")
print("-"*70)
for i in range(min(20, len(R_loc))):
    print(f"{i:<8} {R_loc[i]:<15.4f} {R_amp[i]:<15.6f}")

print(f"\n... (total {len(R_loc)} peaks)")
```

## Respiration Rate Estimation

```python
# Respiration Rate Estimation from EDR Signal
# Based on the methodology: detecting relevant peaks in 60-second windows

# Define window parameters
window_duration = 60  # seconds
sampling_period = time_signal[1] - time_signal[0]

# Calculate the number of samples per window
samples_per_window = int(window_duration / sampling_period)

# Determine total number of windows
total_edr_duration = edr_time[-1] - edr_time[0]
num_windows = int(np.ceil(total_edr_duration / window_duration))

print("="*70)
print("RESPIRATION RATE ESTIMATION FROM EDR SIGNAL")
print("="*70)
print(f"Window duration: {window_duration} seconds")
print(f"Sampling period: {sampling_period:.4f} seconds")
print(f"Samples per window: {samples_per_window}")
print(f"Total EDR duration: {total_edr_duration:.2f} seconds")
print(f"Number of windows: {num_windows}\n")

# Store results for each window
br_results = []

# Process each window
for w in range(num_windows):
    # Define window time range
    window_start_time = edr_time[0] + w * window_duration
    window_end_time = min(window_start_time + window_duration, edr_time[-1])
    
    # Extract window from EDR signal
    window_mask = (edr_time >= window_start_time) & (edr_time < window_end_time)
    xrs_t = edr_signal[window_mask]  # xrs(t) - EDR signal window
    window_time = edr_time[window_mask]
    
    if len(xrs_t) < 100:  # Skip if window is too short
        continue
    
    # Step 1: Find all peaks in the window to determine prominence range
    all_peaks, all_properties = find_peaks(xrs_t, prominence=0)
    
    if len(all_peaks) == 0:
        continue
    
    # Get prominence values for all peaks
    prominences = all_properties['prominences']
    max_pro = np.max(prominences)
    min_pro = np.min(prominences)
    
    # Calculate mean absolute value of the window
    mean_abs_value = np.mean(np.abs(xrs_t))
    
    # Step 2: Identify valid peaks based on three conditions
    # Condition 1: Minimum peak distance = 1.17 seconds (for moderate exercise)
    min_peak_distance_samples = int(1.17 / sampling_period)
    
    # Condition 2: Peak height threshold = 0.95 * mean absolute value
    height_threshold = 0.95 * mean_abs_value
    
    # Condition 3: Prominence threshold = 7.5% of (max_pro - min_pro)
    prominence_threshold = 0.075 * (max_pro - min_pro)
    
    # Find peaks with all three conditions
    valid_peaks, valid_properties = find_peaks(
        xrs_t,
        distance=min_peak_distance_samples,
        height=height_threshold,
        prominence=prominence_threshold
    )
    
    # Calculate breathing rate for this window
    num_breaths = len(valid_peaks)
    
    # BR_edr = (Number of relevant peaks × 60) / window_duration
    BR_edr = (num_breaths * 60) / window_duration
    
    # Calculate true BR from annotated breaths
    # Convert window time to sample indices
    window_start_sample = int(window_start_time * sampling_freq)
    window_end_sample = int(window_end_time * sampling_freq)
    
    # Count annotated breaths within this window
    num_true_breaths = np.sum((breath_samples >= window_start_sample) & (breath_samples <= window_end_sample))
    BR_true = (num_true_breaths * 60) / window_duration
    
    # Calculate error
    error = abs(BR_edr - BR_true)
    error_percent = (error / BR_true * 100) if BR_true > 0 else 0
    
    # Store results
    br_results.append({
        'window': w + 1,
        'start_time': window_start_time,
        'end_time': window_end_time,
        'num_breaths_edr': num_breaths,
        'BR_edr': BR_edr,
        'num_breaths_true': num_true_breaths,
        'BR_true': BR_true,
        'error': error,
        'error_percent': error_percent,
        'max_pro': max_pro,
        'min_pro': min_pro,
        'prominence_threshold': prominence_threshold,
        'height_threshold': height_threshold
    })
    
    print(f"Window {w+1}: {window_start_time:.1f}s - {window_end_time:.1f}s")
    print(f"  Max prominence: {max_pro:.4f}, Min prominence: {min_pro:.4f}")
    print(f"  Prominence threshold: {prominence_threshold:.4f}")
    print(f"  Height threshold: {height_threshold:.4f}")
    print(f"  Valid peaks detected: {num_breaths}")
    print(f"  BR_edr: {BR_edr:.3f} bpm")
    print(f"  BR_true (annotated): {BR_true:.3f} bpm")
    print(f"  Error: {error:.3f} bpm ({error_percent:.2f}%)\n")

# Create DataFrame for analysis
br_df = pd.DataFrame(br_results)

print("\n" + "="*70)
print("BREATHING RATE ESTIMATION SUMMARY")
print("="*70)
print(f"\nMean BR_edr: {br_df['BR_edr'].mean():.3f} bpm")
print(f"Mean BR_true: {br_df['BR_true'].mean():.3f} bpm")
print(f"Mean Absolute Error: {br_df['error'].mean():.3f} bpm")
print(f"Mean Error Percentage: {br_df['error_percent'].mean():.2f}%")
print(f"Std Error: {br_df['error'].std():.3f} bpm")
print(f"Max Error: {br_df['error'].max():.3f} bpm")
print(f"Min Error: {br_df['error'].min():.3f} bpm")

# Visualize results
fig = make_subplots(
    rows=3, cols=1,
    subplot_titles=(
        'EDR Signal with Detected Breath Peaks (60s Windows)',
        'True Respiration Signal with Annotated Breaths',
        'Breathing Rate Comparison: BR_edr vs BR_true'
    ),
    vertical_spacing=0.08
)

# Plot 1: EDR signal with detected peaks per window
fig.add_trace(go.Scatter(
    x=edr_time,
    y=edr_signal,
    mode='lines',
    name='EDR Signal',
    line=dict(color='green', width=1)
), row=1, col=1)

# Add detected peaks for each window
colors_br = ['red', 'blue', 'orange', 'purple', 'brown', 'pink', 'gray', 'olive', 'cyan', 'magenta']

for idx, result in enumerate(br_results):
    window_start = result['start_time']
    window_end = result['end_time']
    
    window_mask = (edr_time >= window_start) & (edr_time < window_end)
    xrs_t = edr_signal[window_mask]
    window_time_plot = edr_time[window_mask]
    
    if len(xrs_t) > 0:
        min_peak_distance_samples = int(1.17 / sampling_period)
        height_threshold = result['height_threshold']
        prominence_threshold = result['prominence_threshold']
        
        valid_peaks, _ = find_peaks(
            xrs_t,
            distance=min_peak_distance_samples,
            height=height_threshold,
            prominence=prominence_threshold
        )
        
        if len(valid_peaks) > 0:
            fig.add_trace(go.Scatter(
                x=window_time_plot[valid_peaks],
                y=xrs_t[valid_peaks],
                mode='markers',
                name=f'Window {idx+1} Peaks',
                marker=dict(color=colors_br[idx % len(colors_br)], size=8, symbol='x'),
                showlegend=(idx < 3)
            ), row=1, col=1)

# Add window boundaries
for i in range(1, num_windows):
    boundary_time = edr_time[0] + i * window_duration
    fig.add_vline(x=boundary_time, line_dash="dash", line_color="gray", opacity=0.4, row=1, col=1)

# Plot 2: True respiration signal with annotated breaths
fig.add_trace(go.Scatter(
    x=time_signal,
    y=resp_signal,
    mode='lines',
    name='True RESP',
    line=dict(color='blue', width=1)
), row=2, col=1)

# Add annotated breaths
fig.add_trace(go.Scatter(
    x=valid_breath_times,
    y=breath_signal_values,
    mode='markers',
    name='Annotated Breaths',
    marker=dict(color='red', size=8, symbol='x')
), row=2, col=1)

# Add window boundaries
for i in range(1, num_windows):
    boundary_time = edr_time[0] + i * window_duration
    fig.add_vline(x=boundary_time, line_dash="dash", line_color="gray", opacity=0.4, row=2, col=1)

# Plot 3: Breathing rate comparison
window_centers = [(r['start_time'] + r['end_time']) / 2 for r in br_results]

fig.add_trace(go.Scatter(
    x=window_centers,
    y=br_df['BR_edr'],
    mode='lines+markers',
    name='BR_edr (Estimated)',
    line=dict(color='green', width=2),
    marker=dict(size=8, symbol='square')
), row=3, col=1)

fig.add_trace(go.Scatter(
    x=window_centers,
    y=br_df['BR_true'],
    mode='lines+markers',
    name='BR_true (Annotated)',
    line=dict(color='blue', width=2, dash='dash'),
    marker=dict(size=8, symbol='circle')
), row=3, col=1)

# Update layout
fig.update_xaxes(title_text='Time [s]', row=3, col=1)
fig.update_yaxes(title_text='Amplitude', row=1, col=1)
fig.update_yaxes(title_text='Amplitude', row=2, col=1)
fig.update_yaxes(title_text='Breathing Rate [bpm]', row=3, col=1)

fig.update_layout(
    height=1200,
    title_text='Respiration Rate Estimation from EDR Signal (Using Annotated Ground Truth)',
    showlegend=True
)

fig.show()

# Display detailed results table
print("\n" + "="*70)
print("DETAILED WINDOW RESULTS")
print("="*70)
print(br_df[['window', 'start_time', 'end_time', 'BR_edr', 'BR_true', 'error', 'error_percent']].to_string(index=False))
```
